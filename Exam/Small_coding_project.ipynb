Great! Now on to removing the punctuation. There isn't (to my knowledge) something like .lower but for removing punctuation. So I'm going to remove the punctuation by doing a for loop. I want Python to look at all the characters in the text and if they are a form of punctuation, then remove it. 

```python
# Instead of having to write out all punctuation types, I import a string module that
# contains all the punctuation characters

import string
string.punctuation
```

```python
# See? Cool. Next: making a for loop that checks every character in the text. This for loop asks:
# Is this character not present in the punctuation string, then add it to the cleaned version. If not, skip it.

text = "test! test test Test test."
cleaned = ""
for char in text:
    if char not in string.punctuation:
        cleaned = cleaned + char

print(cleaned)
```

Yesss, it works! Now on to combining them to one function. 

```python
text = "test! test test Test test."

def clean_up(text):
    lowered = text.lower()
    cleaned = ""
    for char in lowered: 
        if char not in string.punctuation:
            cleaned = cleaned + char
    return cleaned

print(clean_up(text))
```
